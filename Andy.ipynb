```python
## Year and GDP distribution
```

```python
years = [1950, 1960, 1970, 1980, 1990, 2000, 2010]
gdp = [300.2, 543.3, 1075.9, 2862.5, 5979.6, 10289.7, 14958.3]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


plt.figure(figsize=(10, 6))  # Adjust the figure size if needed

# Use a bar chart with width adjusted for better visualization
bar_width = 4  # Adjust the width as needed
plt.bar(years, gdp, width=bar_width, color='skyblue')

plt.xlabel('Year')
plt.ylabel('GDP')
plt.title('Year and GDP Distribution')

plt.show()
```

**BAR CHART**

```python
plt.bar(range(len(years)), gdp)
plt.title("Year and GDP Distribution")
plt.ylabel("GDP")
plt.xticks(range(len(years)), years)
plt.show()
```

LINE CHART

```python
plt.figure(figsize=(10, 6))  # Adjust the figure size if needed

# Plot the line chart
plt.plot(years, gdp, marker='o', linestyle='-', color='red')  # 'o' for circles, '-' for solid line

plt.xlabel('Year')
plt.ylabel('GDP Billion of $USD')
plt.title('Nominal GDP')

plt.grid(True)  # Add a grid for better readability

plt.show()
```

**SHOW DIAGRAM**


```python
variance = [1, 2, 4, 8, 16, 32, 64, 128, 256]
bias_squared = [256, 128, 64, 32, 16, 8, 4, 2, 1]
total_error = [v + b for v, b in zip(variance, bias_squared)]
```

```python
# Tạo danh sách các chỉ số cho trục x
x = list(range(len(variance)))

# Vẽ các đường biểu diễn
plt.plot(x, variance, label='variance',  color='green', linewidth=1)
plt.plot(x, bias_squared, label='bias^2', linestyle='-.', color='red',linewidth=1)
plt.plot(x, total_error, label='total error',linestyle=':', color='blue',linewidth=0.85)

# Thêm tiêu đề và nhãn
plt.title('Bias-Variance Tradeoff')
plt.xlabel('Model Complexity')
plt.ylabel('Error')

# Hiển thị chú thích
plt.legend()

# Hiển thị biểu đồ
plt.show()
```
